```python
import pandas as pd
```

```python
dataset = pd.read_excel('./pesquisa.xlsx', header=1)
dataset.columns.values
```

```python
dataset[['Tipo', 'Tipo Endereço','Zoneamento Novo', 'Bairro', 'Cidade', 'Formato', 'Topografia', 'Característica da construção']] = dataset[['Tipo', 'Tipo Endereço','Zoneamento Novo', 'Bairro', 'Cidade', 'Formato', 'Topografia', 'Característica da construção']].apply(lambda x: x.astype('category').cat.codes)
dataset[['Tipo', 'Tipo Endereço','Zoneamento Novo', 'Bairro', 'Cidade', 'Formato', 'Topografia', 'Característica da construção']]
```

```python
dataset = dataset.drop(columns=['estado'])
```

```python
x = dataset['Melhoramentos Públicos']
myset = set()
for phrase in x:
    for word in phrase.split(", "):
        word.removesuffix(", ").removesuffix(",")
        print(word)
        myset.add(word)

myset.discard('Rede de Esgoto,')
myset.discard("")
myset.discard(" ")

print(myset)
myset
```

```python

catdf = pd.DataFrame(columns=list(myset) )
listofsets = []
for phrase in x:
    myset.clear()
    for word in phrase.split(", "):
        word.removesuffix(", ").removesuffix(",")
        myset.add(word)
    myset.discard('Rede de Esgoto,')
    myset.discard("")
    myset.discard(" ")
    listofsets.append(myset)

listofsets
```

```python
catdf = pd.DataFrame(listofsets, columns=catdf.columns)
catdf
```
